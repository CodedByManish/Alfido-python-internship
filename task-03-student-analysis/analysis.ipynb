```python
import pandas as pd

# Load dataset
df = pd.read_csv("students.csv")
print(f"Dataset loaded.")
```

```python
# Check structure and missing values
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Fill missing marks with mean value
mean_marks = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(mean_marks)

print(f"Missing values filled with mean: {round(mean_marks, 1)}")

df
```

```python
# Students scoring above 90
top_students = df[df["Marks"] > 90]

print(f"Top students count: {len(top_students)}")
top_students
```

```python
# Department-wise average marks
dept_summary = df.groupby("Department")["Marks"].mean().round(1).reset_index()

print("Department-wise average marks:")
dept_summary
```

```python
# Overall statistics
df["Marks"].describe().round(1)
```
